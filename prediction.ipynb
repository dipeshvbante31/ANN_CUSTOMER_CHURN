```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle 
import pandas as pd 
import numpy as np 
```

```python
#load all train model 
model = load_model('model.h5')
#Load encoder and scaler 
with open('ohe_geo.pkl','rb') as file:
    lable_encoder_geo=pickle.load(file)

with open('ohe_gen.pkl','rb') as file:
    lable_encoder_gen=pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
geo_encoded = lable_encoder_geo.transform([[input_data['Geography']]])
geo_df=pd.DataFrame(geo_encoded,columns=lable_encoder_geo.get_feature_names_out(['Geography']))
```

```python
geo_df
```

```python
import pandas as pd

input_data = pd.DataFrame([input_data])

input_data = pd.concat(
    [input_data.reset_index(drop=True), geo_df.reset_index(drop=True)],
    axis=1
)
```

```python
input_data
```

```python
input_data['Gender']=lable_encoder_gen.transform(input_data['Gender'])
```

```python
input_data
```

```python
input_data.drop(columns=["Geography"], inplace=True)
```

```python
# concatination with one hot encoded data 
# input_data=pd.concat([input_data.drop("Geography",axis=1),geo_df],axis=1)
```

```python
input_data
```

```python
#scaling the input data 
input_scaled=scaler.transform(input_data)
```

```python
input_scaled 
```

```python
#predict churn 
prediction=model.predict(input_scaled)
print(100*prediction)
```

```python
prediction_proba=prediction[0][0]
```

```python
if prediction_proba > 0.5:
    print("the customer is likely to churn")
else:
    print("the customer is not likely to churn")
```
