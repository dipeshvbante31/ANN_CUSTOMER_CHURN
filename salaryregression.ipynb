```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler,OneHotEncoder
import pickle
```

```python
data =pd.read_csv('churn.csv')
```

```python
data.head()
```

```python
data = data.drop(['RowNumber','CustomerId','Surname'],axis=1)
```

```python
data 
```

```python
label_gen=LabelEncoder()
data['Gender']=label_gen.fit_transform(data['Gender'])
```

```python
data
```

```python
ohe_geo=OneHotEncoder()
geo_encoded=ohe_geo.fit_transform(data[['Geography']]).toarray()
geo_df=pd.DataFrame(geo_encoded,columns=ohe_geo.get_feature_names_out(['Geography']))
```

```python
data=pd.concat([data.drop('Geography',axis=1),geo_df],axis=1)
```

```python
data
```

```python
#splitting the data into independent and dependent features 
x=data.drop('EstimatedSalary',axis=1)
y=data['EstimatedSalary']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
with open('label_gen.pkl','wb') as file:
    pickle.dump(label_gen,file)
    
with open('geo.pkl','wb') as file:
    pickle.dump(ohe_geo,file)

with open('stdscaler2.pkl','wb') as file:
    pickle.dump(scaler,file)
```

## ANN regression problem statement 

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import tensorflow as tf 

```

```python
model = Sequential([
    Dense(64,activation='relu',input_shape=(x_train.shape[1],)),
    Dense(32,activation='relu'),
    Dense(1)# output layer for regression 
])
```

```python
model.compile(optimizer='adam',loss='mean_absolute_error',metrics=['mae'])
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime

log_dir = "regressionlogs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")

tb_callback = TensorBoard(
    log_dir=log_dir,
    histogram_freq=1
)
```

```python
# early stopping setup 
early_stopping_callbacks= EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
history  = model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=100,callbacks=[early_stopping_callbacks,tb_callback])
```

```python
#evaluate the model for test data 
test_loss,test_mae=model.evaluate(x_test,y_test)
print(f"test mae : {test_mae}")
print(f"test loss : {test_loss}")
```
